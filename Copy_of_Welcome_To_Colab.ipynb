```
import pandas as pd

data = pd.read_csv("ipl2024 Matches.csv")

while True:
    print("\n--- IPL 2024 ---")
    print("1. Find match winner")
    print("2. Find toss winner")
    print("3. Find Player of the Match")
    print("4. Show matches of a team")
    print("5. Count team wins")
    print("6. Show highest score")
    print("7. Total matches")
    print("8. Exit")

    choice = input("Enter your choice: ")

    if choice == "1" or choice == "2" or choice == "3":
        mid = int(input("Enter match ID: "))
        found = False

        for i in range(len(data)):
            if data.loc[i, "id"] == mid:
                if choice == "1":
                    print("Winner:", data.loc[i, "winner"])
                elif choice == "2":
                    print("Toss winner:", data.loc[i, "toss_winner"])
                else:
                    print("Player of the Match:", data.loc[i, "player_of_the_match"])
                found = True
                break

        if found == False:
            print("Match not found")

    elif choice == "4":
        team = input("Enter team name: ").lower()
        found = False

        for i in range(len(data)):
            if team == data.loc[i, "team1"].lower() or team == data.loc[i, "team2"].lower():
                print(data.loc[i, "team1"], "vs", data.loc[i, "team2"],
                      "- Winner:", data.loc[i, "winner"])
                found = True

        if found == False:
            print("Team not found")

    elif choice == "5":
        team = input("Enter team name: ").lower()
        wins = 0

        for i in range(len(data)):
            if data.loc[i, "winner"].lower() == team:
                wins = wins + 1

        print("Total wins:", wins)

    elif choice == "6":
        score = data["first_score"].max()
        print("Highest first innings score:", score)

    elif choice == "7":
        print("Total matches:", len(data))

    elif choice == "8":
        print("Thank you")
        break
```
